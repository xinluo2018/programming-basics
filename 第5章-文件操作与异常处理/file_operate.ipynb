### 文件操作与异常处理

```python
f = open('第5章-文件操作与异常处理/file.txt', 'r')
content = f.read()
print(content)
f.close()  # 关闭文件
```

```python
## with...as...上下文管理，自动关闭文件
with open('第5章-文件操作与异常处理/file.txt', 'r') as file:
    content = file.read()
print(content)
```

```python
## 写出
with open('第5章-文件操作与异常处理/file_2.txt', 'w') as f:
    f.write('Hello, world!\n')
    f.write('This is a new file.\n')
```

```python
## 追加
with open('第5章-文件操作与异常处理/file_2.txt', 'a') as f:
    f.write('This is a sentence using "a" mode.\n')
```

##### 程序有异常，但不希望中断：异常捕获

```python
number = int(input("请输入整数："))
print("输入的整数是：", number)
```

```python
try:
    number = int(input("请输入整数："))
    print("输入的整数是：", number)
except ValueError:
    print("输入无效，请输入整数")
```

##### 捕获多个异常

```python
try:
  num = int(input("请输入: "))
  result = 10 / num
  print(result)
except ZeroDivisionError:
  print(f"除数不能为零")
```

```python
## 捕获多个异常
try:
  num = int(input("请输入: "))
  print(10/num)
except ZeroDivisionError:
  print("除数不能为零")
except ValueError:
  print("输入的不是一个数字")
```

#### assert: 断言

```python
def calculate_rect_area(width, height):
    return width * height

area = calculate_rect_area(4,5)
```

```python
### 验证程序内部假设
def calculate_rect_area(width, height):
    assert width > 0, "宽度必须为正数"
    assert height > 0, "高度必须为正数"
    return width * height

# print(calculate_rect_area(5, 3))  # 正常
print(calculate_rect_area(-2, 4))   # 触发断言 
```
